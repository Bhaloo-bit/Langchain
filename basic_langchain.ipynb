```python
import os
from langchain.chat_models import init_chat_model
llm = init_chat_model(model=os.getenv("MODEL", "gpt-6-luna"),
                      model_provider=os.getenv("PROVIDER", "openai"))

res = llm.invoke([
    {"role":"system", "content":"You are a help full person"}, # system prompt setting the behaviour of the model
    {"role":"user", "content":"how to find second largest element in the array"} # user promt QnA
    
])

print(res.content)
```

other method for messeging

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
promts =[
    SystemMessage(content="your are help full ai assitance"),
    HumanMessage(content="hi would we my code help agents"),
    AIMessage(content="response")
]
res= llm.invoke(promts)
print(res.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"you are a senior develop in {language}"},
    {"role":"user", "content":"{query}"}
])


```

```python
prompt
```

## LCEL | operator 
 user_requrest -> promot -> llm -> response

```python
question = "how to map the array"
language = "javascripts"
finalprompt = prompt.invoke({"language": language, "query": question})

res = llm.invoke(finalprompt)
```

```python
print(res.content)
```

### LangChain Chain


```python
from langchain.messages import AIMessage
def getContent(llmResponse:AIMessage):
    return llmResponse.content

def shortTheMessage(msg:str):
    return msg[:100]
```

```python
qna_chain = prompt | llm | getContent | shortTheMessage
```

```python
res = qna_chain.invoke({"language": "english", "query": "what is gen ai"})
```

```python
print(res)
```
